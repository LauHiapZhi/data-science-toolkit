```python
import numpy as np

array_3x4 = np.random.rand(3,4)

print("The random 3x4 array is ")
print(array_3x4)
```

```python
import numpy as np

arrayof1 = np.ones((3,3))
arrayof0 = np.zeros((3,3))

print("Ones array is")
print(arrayof1)
print("Zeros array is")
print(arrayof0)
```

```python
import numpy as np

array_6x3 = np.random.rand(6,3)
transposed_array = np.transpose(array_6x3)

print("Original array is")
print(array_6x3)
print("Transposed array is ")
print(transposed_array)
```

```python
import numpy as np

print("Enter the number of n, p and m for nxp array and pxm array: ")
n =int (input("Number of n "))
p =int (input("Number of p "))
m =int (input("Number of m "))

array_nxp = np.random.rand(n,p)
array_pxm = np.random.rand(p,m)
dotproduct = np.dot(array_nxp,array_pxm)

print("Matrix A is")
print(array_nxp)
print("Matrix B is")
print(array_pxm)

print("The dot product of matrix A and B is")
print(dotproduct)
```
